```python
import pandas as pd
import geopandas as gpd
import json
from shapely.geometry import shape
from shapely import wkt
import warnings
from shapely.ops import linemerge, unary_union

# filter all warnings
warnings.filterwarnings("ignore")
```

```python
zona_u = shape({
    'type': 'Point',
    'coordinates': (2.114055470466503, 41.38275273204058)
})

mundet = shape({
    'type': 'Point',
    'coordinates': (2.1455911047955754, 41.43508184671605)
}) 

poble_nou = shape({
    'type': 'Point',
    'coordinates': (2.198402153798261, 41.40288229103046)
}) # poble nou

destinations = [zona_u, mundet, poble_nou]
destinations = gpd.GeoDataFrame(geometry=destinations, crs="EPSG:4326")
destinations['name'] = ['Zona Universitària', 'Mundet', 'Poble Nou']
destinations['demand'] = [1000,500, 200]
destinations = destinations[['name', 'demand', 'geometry']]
destinations.to_csv("Nodes/Destinations.csv", index=False)
```
