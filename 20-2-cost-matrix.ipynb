```python
to_dest = pd.read_csv('Data/Djisktra-out/export.csv')

to_dest = to_dest.drop_duplicates(subset=['route_id'],keep = 'first')
to_dest = to_dest[['from','to','totalCost']].rename(columns={'from':'origen','to':'dest','totalCost':'cost'})
to_dest_top100 = (
    to_dest
    .drop_duplicates(subset="route_id", keep="first")
    .sort_values(["dest", "cost"])
    .groupby("dest", group_keys=False)
    .head(100)
)
iu_bus = pd.read_csv('Edges/IU-Bus.csv',low_memory=False)
iu_bus = iu_bus[['origen','dest','time','length']]
iu_bus['length'] = iu_bus['length']/1000
iu_bus.rename(columns={'time':'cost','length':'distance'}, inplace=True)
cost_matrix = pd.concat([to_dest_top100, iu_bus], ignore_index=True)
cost_matrix.to_csv('Data/Feed_optimization/Cost_matrix.csv', index=False)
```
