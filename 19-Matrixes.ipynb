```python
import pandas as pd
import glob
```

```python
to_dest = pd.concat(
    [pd.read_csv(f) for f in sorted(glob.glob(f"Data/Djisktra-out/export_*.csv"))],
    ignore_index=True)

to_dest = to_dest.drop_duplicates(subset=['route_id'],keep = 'first')
to_dest = to_dest[['from','to','totalCost']].rename(columns={'from':'origen','to':'dest','totalCost':'cost'})
iu_bus = pd.read_csv('Edges/IU-Bus.csv',low_memory=False)
iu_bus = iu_bus[['origen','dest','time','length']]
iu_bus['length'] = iu_bus['length']/1000
iu_bus.rename(columns={'time':'cost','length':'distance'}, inplace=True)
cost_matrix = pd.concat([to_dest, iu_bus], ignore_index=True)
cost_matrix.to_csv('Data/Feed_optimization/Cost_matrix.csv', index=False)
```

```python
demand_matrix = pd.read_csv('Nodes/N-POIs.csv')

demand_matrix = (
    demand_matrix
    .groupby('poi_name', as_index=False)['demand']
    .sum()
    .set_index('poi_name')
    .T
    .reset_index(drop=True)
)

demand_matrix.to_csv('Data/Feed_optimization/Demand_matrix.csv', index=False)
```
